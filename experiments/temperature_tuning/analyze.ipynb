```python
import pickle 
import numpy as np
from typing import List
import matplotlib.pyplot as plt
import os
from pyparsing import Word, alphas, Literal, Combine
from matplotlib import font_manager
font_path = "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc"
font_prop = font_manager.FontProperties(fname=font_path)
```

```python
def load_pkl(cache_file_path: str = None):
    with open(cache_file_path, 'rb') as f:
        loaded_data = pickle.load(f)
    return loaded_data

# cache_file_path = "/root/llm_inference_boilerplate_code/cache/math_0_2500_qwen2.5-math-7b-instruct.pkl"
# data = load_pkl(cache_file_path)

def load_pkls(cache_file_paths: List[str] = None):
    data = None 
    for cache_file_path in cache_file_paths:
        print("loading file: ", cache_file_path)
        with open(cache_file_path, 'rb') as f:
            loaded_data = pickle.load(f)
        
        if data is None:
            data = loaded_data.copy()
        else:
            for k in data.keys():
                data[k].extend(loaded_data[k]) 
    return data
```

```python
pkls = [
    "/root/llm_inference_boilerplate_code/cache/math_0_2500_qwen2.5-math-7b-instruct.pkl",
    "/root/llm_inference_boilerplate_code/cache/math_2500_5000_qwen2.5-math-7b-instruct.pkl"
]
data = load_pkls(pkls)
```

```python
def plot_token_matrix(results_per_temp, save_folder, max_cols=50, top_n=3, qid=0):
    os.makedirs(save_folder, exist_ok=True)

    for T, top3_info in results_per_temp.items():
        n_steps = len(top3_info)
        max_cols = min(max_cols, n_steps)
        n_blocks = (n_steps + max_cols - 1) // max_cols
        total_rows = n_blocks * top_n
        total_cols = max_cols

        prob_matrix = np.zeros((total_rows, total_cols))
        token_matrix = [["" for _ in range(total_cols)] for _ in range(total_rows)]

        step_index = 0
        for block_i in range(n_blocks):
            block_start = block_i * max_cols
            block_end = min(block_start + max_cols, n_steps)
            block_size = block_end - block_start

            for c in range(block_size):
                topk = top3_info[step_index]
                step_index += 1
                for rank in range(top_n):
                    row = block_i * top_n + rank
                    col = c
                    token_str, prob = topk[rank]
                    prob_matrix[row, col] = prob
                    token_matrix[row][col] = token_str

        plt.figure(figsize=(total_cols * 1.2, total_rows * 1.2))
        plt.imshow(prob_matrix, aspect='auto')
        plt.colorbar(label='Probability')
        plt.title(f"Heatmap of Top-{top_n} Tokens (T={T})")

        for r in range(total_rows):
            for c in range(total_cols):
                tok = token_matrix[r][c]
                tok = tok.replace(r'$', r'\$')
                if tok:
                    val = prob_matrix[r, c].item()
                    plt.text(c, r, f"{tok}\n({val:.2f})", ha='center', va='center', fontsize=8, color='red')

        plt.xlabel("Decoding Step")
        plt.ylabel("Row = block * 3 + rank\n(0=top1, 1=top2, 2=top3)")
        save_path = os.path.join(save_folder, f"token_heatmap_{qid}.pdf")
        plt.savefig(save_path, bbox_inches='tight')
        plt.close()
```

```python
topk_level = 3
qid=5
font_path = "/usr/share/fonts/opentype/noto/NotoSansCJK-Bold.ttc"
font_prop = font_manager.FontProperties(fname=font_path)
# for qid in range(len(data['responses'])):
results = []
for time_step in range(data['topk_probs'][qid].shape[0]):
    results.append([(data['topk_tokens'][qid][time_step][k],
                    data['topk_probs'][qid][time_step, k]) for k in range(topk_level)])
    if data['topk_tokens'][qid][time_step][0] == "<|im_end|>":
        break
plot_token_matrix({1.0: results}, ".", max_cols=50, top_n=3, qid=qid)
```
